```python
# Exercise 1: Emergency Dispatch Record Search

# 1. Student Identity Configuration
LAST_NAME = "AMISTAD"
STUDENT_ID = "TUPM-26-1938"
FAVORITE_ARTIST = "THE BEATLES"

# 2. Derive Student-Specific Inputs
SEED_DIGIT = int(STUDENT_ID[-1])                                   
ID_SUM = sum(int(d) for d in STUDENT_ID if d.isdigit())            
NAME_LENGTH = len(LAST_NAME)                                       
ARTIST_LENGTH = len(FAVORITE_ARTIST)                               
ARTIST_SUM = sum(ord(c.upper()) - 64 for c in FAVORITE_ARTIST if c.isalpha()) 

# 3. Sorted Original Dataset
dataset_size = 15
original_records = sorted(list(set([(ID_SUM * i + 7) for i in range(1, dataset_size + 1)])))

# 4. Rotate Dataset
rotation_index = SEED_DIGIT % len(original_records)
rotated_records = original_records[rotation_index:] + original_records[:rotation_index]

# 5. Generate Target Incident
target_incident = ID_SUM + ARTIST_SUM  

if target_incident not in rotated_records:
    rotated_records[5] = target_incident

# 6. Binary Search
def search_rotated_array(arr, target):
    low, high = 0, len(arr) - 1
    comparisons = 0
    
    while low <= high:
        mid = (low + high) // 2
        comparisons += 1
        
        if arr[mid] == target:
            return mid, comparisons
        
        if arr[low] <= arr[mid]:
            comparisons += 1
            if arr[low] <= target < arr[mid]:
                high = mid - 1
            else:
                low = mid + 1
        else:
            comparisons += 1
            if arr[mid] < target <= arr[high]:
                low = mid + 1
            else:
                high = mid - 1
                
    return -1, comparisons

# 7. Linear Search
def linear_search(arr, target):
    comparisons = 0
    for i, val in enumerate(arr):
        comparisons += 1
        if val == target:
            return i, comparisons
    return -1, comparisons

rot_index, rot_comparisons = search_rotated_array(rotated_records, target_incident)
lin_index, lin_comparisons = linear_search(rotated_records, target_incident)

print("=" * 90)
print("  EXERCISE 1: EMERGENCY DISPATCH RECORD SEARCH REPORT")
print("=" * 90)
print(f"Inputs: LAST_NAME= {LAST_NAME} | SEED_DIGIT = {SEED_DIGIT} | ARTIST = {FAVORITE_ARTIST}")
print(f"Derived Metrics: ID_SUM = {ID_SUM} | ARTIST_SUM = {ARTIST_SUM}")
print("-" * 90)
print(f"Original Dataset : {original_records}")
print(f"Rotated Dataset  : {rotated_records}")
print(f"Target Incident  : {target_incident}")
print("-" * 90)
print(f"Binary Search -> Index: {rot_index} | Comparisons: {rot_comparisons}")
print(f"Linear Search -> Index: {lin_index} | Comparisons: {lin_comparisons}")
print("=" * 90)
```

```python
# Exercise 2: University Enrollment Records

# 1. Student Identity Configuration
LAST_NAME = "AMISTAD"
STUDENT_ID = "TUPM-26-1938"
FAVORITE_ARTIST = "THE BEATLES"

# 2. Derive Student-Specific Inputs
SEED_DIGIT = int(STUDENT_ID[-1])                                  
ID_SUM = sum(int(d) for d in STUDENT_ID if d.isdigit())            
NAME_LENGTH = len(LAST_NAME)                                      
ARTIST_LENGTH = len(FAVORITE_ARTIST)                               

# 3. Generate Sorted Enrollment Datasets
# Dept A
dept_a = sorted(list(set([(ID_SUM * i + SEED_DIGIT) for i in range(1, 8)])))

# Dept B
dept_b = sorted(list(set([(ARTIST_LENGTH * i + NAME_LENGTH) for i in range(1, 9)])))

# 4. Consolidate Two Sorted Records
def merge_sorted_records(arr1, arr2):
    consolidated = []
    i = 0
    j = 0
    comparisons = 0
    
    while i < len(arr1) and j < len(arr2):
        comparisons += 1
        if arr1[i] <= arr2[j]:
            consolidated.append(arr1[i])
            i += 1
        else:
            consolidated.append(arr2[j])
            j += 1
            
    while i < len(arr1):
        consolidated.append(arr1[i])
        i += 1
        
    while j < len(arr2):
        consolidated.append(arr2[j])
        j += 1
        
    return consolidated, comparisons

consolidated_dataset, total_comparisons = merge_sorted_records(dept_a, dept_b)
is_sorted = consolidated_dataset == sorted(consolidated_dataset)

print("=" * 90)
print("  EXERCISE 2: UNIVERSITY ENROLLMENT RECORDS REPORT")
print("=" * 90)
print(f"Inputs: SEED_DIGIT={SEED_DIGIT} | ID_SUM={ID_SUM} | NAME_LEN={NAME_LENGTH} | ARTIST_LEN={ARTIST_LENGTH}")
print("-" * 90)
print(f"Department A Dataset : {dept_a}")
print(f"Department B Dataset : {dept_b}")
print(f"Consolidated Dataset : {consolidated_dataset}")
print("-" * 90)
print(f"Records Count Before Consolidation : {len(dept_a)} + {len(dept_b)} = {len(dept_a) + len(dept_b)}")
print(f"Records Count After Consolidation  : {len(consolidated_dataset)}")
print(f"Total Comparisons Performed        : {total_comparisons}")
print(f"Ordering Verification              : {'Sorted' if is_sorted else 'FAILED'}")
print("=" * 90)
```

```python
# Appendix Exercise 3: E-Commerce Order Processing

import time
import random

# 1. Student Identity Configuration
LAST_NAME = "AMISTAD"
STUDENT_ID = "TUPM-26-1938"
FAVORITE_ARTIST = "THE BEATLES"

# 2. Derive Student-Specific Inputs
SEED_DIGIT = int(STUDENT_ID[-1])                                   
ID_SUM = sum(int(d) for d in STUDENT_ID if d.isdigit())            
NAME_LENGTH = len(LAST_NAME)                                       
ARTIST_LENGTH = len(FAVORITE_ARTIST)                               

# 3. Generate Oiginal Dataset 
base_orders = [(ID_SUM * i + SEED_DIGIT * 3) for i in range(1, 11)]

# Create the three input conditions from the same order set
sorted_input = sorted(base_orders)
reverse_input = sorted(base_orders, reverse=True)

random.seed(SEED_DIGIT)
random_input = base_orders.copy()
random.shuffle(random_input)

# 4. Sorting Algorithms with Tracking

def bubble_sort(arr):
    a = arr.copy()
    n = len(a)
    comparisons, movements = 0, 0
    start = time.perf_counter()
    for i in range(n):
        swapped = False
        for j in range(0, n - i - 1):
            comparisons += 1
            if a[j] > a[j + 1]:
                a[j], a[j + 1] = a[j + 1], a[j]
                movements += 1
                swapped = True
        if not swapped:
            break
    elapsed = time.perf_counter() - start
    return a, comparisons, movements, elapsed

def selection_sort(arr):
    a = arr.copy()
    n = len(a)
    comparisons, movements = 0, 0
    start = time.perf_counter()
    for i in range(n - 1):
        min_idx = i
        for j in range(i + 1, n):
            comparisons += 1
            if a[j] < a[min_idx]:
                min_idx = j
        if min_idx != i:
            a[i], a[min_idx] = a[min_idx], a[i]
            movements += 1
    elapsed = time.perf_counter() - start
    return a, comparisons, movements, elapsed

def insertion_sort(arr):
    a = arr.copy()
    comparisons, movements = 0, 0
    start = time.perf_counter()
    for i in range(1, len(a)):
        key = a[i]
        j = i - 1
        while j >= 0:
            comparisons += 1
            if a[j] > key:
                a[j + 1] = a[j]
                movements += 1
                j -= 1
            else:
                break
        a[j + 1] = key
    elapsed = time.perf_counter() - start
    return a, comparisons, movements, elapsed

inputs = {
    "Sorted": sorted_input,
    "Reverse-Ordered": reverse_input,
    "Random": random_input
}

algorithms = {
    "Bubble Sort": bubble_sort,
    "Selection Sort": selection_sort,
    "Insertion Sort": insertion_sort
}

print("=" * 90)
print("  EXERCISE 3: E-COMMERCE ORDER PROCESSING REPORT")
print("=" * 90)
print(f"Base Orders Dataset: {base_orders}")
print("-" * 90)

for condition_name, dataset in inputs.items():
    print(f"\n--- Input Condition: {condition_name} ---")
    print(f"Data Sequence: {dataset}")
    print(f"{'Algorithm':<18} | {'Comparisons':<12} | {'Movements':<12} | {'Time (s)':<12} | {'Sorted Verification'}")
    print("-" * 85)
    for algo_name, algo_func in algorithms.items():
        res, comp, mov, t = algo_func(dataset)
        is_correct = res == sorted(dataset)
        print(f"{algo_name:<18} | {comp:<12} | {mov:<12} | {t:<12.8f} | {'PASSED' if is_correct else 'FAILED'}")

print("=" * 90)
```
